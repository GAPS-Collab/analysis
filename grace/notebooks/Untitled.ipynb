```python
import gondola
from pathlib import Path
```

```python
tof_data_path = Path("/data1/nextcloud/cra_data/data/2024/live/tof/9125")
files = sorted(tof_data_path.glob("*.tof.gaps"))[-2:]
```

```python
print(files[1])
```

```python
rb_calib_path = Path('/data1/nextcloud/cra_data/data/2024/live/tof/calib/241212_125129UTC')
```

```python
rb_calibrations = gondola.calibration.load_rb_calibrations(rb_calib_path)
```

```python
tof_db = gondola.db.TofPaddle.all_as_dict()
```

```python
n_tof_events = 0
n_waveforms = 0
for f in files:
    reader = gondola.io.TofPacketReader(str(f))
    for packet in reader: 
        if packet.packet_type in [gondola.packets.TofPacketType.TofEvent, gondola.packets.TofPacketType.TofEventDeprecated]:
            tof_event = gondola.events.TofEvent()
            gondola.visual.tof.plot_waveforms(tof_event, calib = rb_calibrations, with_hits=False, skip_bins=0)
```
